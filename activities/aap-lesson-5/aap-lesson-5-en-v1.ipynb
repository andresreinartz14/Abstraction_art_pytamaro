## Reusing a Shape by Changing Its View

In earlier lessons, you broke artworks into parts, gave those parts names, changed them with parameters, and repeated them with purpose.

Now we add a new move: **transformation**.

A transformation changes how a graphic appears without changing the idea behind it. In this lesson, our main transformation is **rotation**.

> Artist brain: “What if this shape faced another direction?”  
> Programmer brain: “What if I call `rotate`?”

## Concept: rotation as abstraction

When programmers abstract, they often ask:

> “What is the one idea hiding inside all these similar cases?”

Here, the one idea is a **right triangle**. Instead of drawing four different triangles manually, we can draw one triangle and rotate it.

That is abstraction because we separate:

- the **base idea**: a right triangle
- the **variation**: which direction it faces
- the **composition**: how rotated copies are pinned together

In art, this is similar to using a repeated motif. In code, it becomes a reusable function plus transformations.

In pyTamaro, each shape has a "bounded box" around it. To get shapes to connect with each other in specific ways,

Enter the concept of pinning!!

## Pinning: choosing how pieces meet

Rotation only turns a shape. To make shapes meet, we need <API>pin</API>-ing. Pinning chooses a meaningful point of a graphic, such as its center or edge, so we can compose it with another graphic.

## Make one square-like structure

So let's recap what this function should do:
