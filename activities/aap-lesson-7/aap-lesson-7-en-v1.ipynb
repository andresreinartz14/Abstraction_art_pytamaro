## Build Your Visual Vocabulary

Artists often begin with a few visual ingredients. Programmers do the same with functions.

<Task>
Run the cell below. Below are a couple "pre-made" building blocks you could use as a start. 

Now add a few of your own! Decide which ingredients **feel** like they belong in your final artwork.
</Task>

```python
def square(size: float, color: Color) -> Graphic:
    return rectangle(size, size, color)


def thin_bar(length: float, thickness: float, color: Color) -> Graphic:
    return rectangle(length, thickness, color)


def dot(size: float, color: Color) -> Graphic:
    return ellipse(size, size, color)


show_graphic(
    beside(
        square(70, PAPER),
        beside(
            thin_bar(70, 12, INK),
            dot(35, RED)
        )
    )
)
```
