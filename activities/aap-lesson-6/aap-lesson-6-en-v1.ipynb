## Concept: Rule-Based Art

A rule-based artwork does not mean “the computer makes all the choices and the artist goes for a snack.”

It means the artist chooses a **system**.

Here are three different systems for coloring a grid:

1. **Fixed pattern**: every third square is blue.
2. **Total randomness**: every square randomly chooses any color.
3. **Controlled randomness**: squares near the center are more likely to be white.

All three can use the same grid. What changes is the **decision rule**.

## Controlled Variation: More White Near the Center

In our improved version, every cell still has some randomness. But the randomness will be **guided** by position.

The idea:

> Cells closer to the center are more likely to be white.

If we look at an overlayed version of the painting, we can clearly see a rough trend of white cells being more prevelant near the center. 

![Composition with Grid IX](white.png)

That is controlled variation. The computer can still surprise us, but the artist has chosen the direction of the surprise.

To do this, we need a function:

```python
white_likelihood(i, j, mid, margin)
```

It will return a number between `0.0` and `1.0`.

- `0.0` means “no special white effect here.”
- `1.0` means “maximum chance of white here.”

## A Tiny Bit of Geometry: Distance from the Center

To know whether a cell is near the center, we measure its distance from the center.

For a cell at row `i` and column `j`, and a center position `mid`, the distance is:

```python
distance = ((i - mid) ** 2 + (j - mid) ** 2) ** 0.5
```

This is the **Pythagorean theorem** in disguise.

You may know it as:

```text
a² + b² = c²
```

Here:

- `i - mid` is the vertical distance from the center,
- `j - mid` is the horizontal distance from the center,
- the final `** 0.5` means “take the square root,” giving us the diagonal distance.

To visualize let's randomly select a cell from the painting. We can see that the cell is not only to the right of the middle of the canvas, but also a bit above it. Essentially, the cell has an x and y component relative to the middle of the canvas. 

We can almost imagine that the center of the canvas is the **origin** of a graph, and if we trace the path from our **origin** to our cell we end up with a pythagoras-like scenario!
From there we just apply the equation to find the distance of our hypotenuse. 

![Composition with Grid IX](pythagoras.png)


So the program is basically asking:

> “How far is this square from the middle of the artwork?”

The closer the answer is to `0`, the more central the square is.

## Why the Controlled Version Feels More Designed

The controlled version still uses randomness, but not everywhere in the same way.

The rule creates a **center-weighted composition**:

- the center has a high chance of white,
- cells farther away have less chance of white,
- beyond the margin, cells have no extra white effect.

That means the artwork has a visual focus without becoming perfectly predictable.
