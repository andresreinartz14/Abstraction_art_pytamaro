## The Real Deal
We've made some solid progress here, but what we were working with was a **simplified** version of the painting. The good news is it's almost the same. The KEY difference is in the actual painting, the right panel is a "special panel".

The special panel is itself composed of three panels, one of which is `BLUE`.

![Vertical Composition with Blue and White](painting2.png)

Clearly the special panel itself also uses proportions, in this case, relative to the total `canvas_height`. The bottom-most panel occupies `1/10th`, the middle panel `7/10th`, and the top panel `2/10th`'s of it.

If you think about it, it's almost the same approach we used for our `composition` function, but this time instead of stacking panels besides each other we're stacking them on top of each other. The key differences are:

1. Now the proportions of the panels are in relation to the `canvas_height` (not the width).
2. We're not just working with `white_panel`s. The top panel is blue.

We could implement this in a method named `special_panel`, but you know what would be even more abstract? A single method named `panel` that handles both the case of a regular `white_panel` and the case where we construct a `special_panel`.

## Build the Full Composition

Believe it or not, we're pretty much already there! We can re-use our past `composition` function, the only difference being, now instead of using the `white_panel` function to create the left, middle, and right panels, we call our newly made `panel` function which handles the special panel case.

```python

def composition(canvas_width: float, canvas_height: float, line: float) -> Graphic:
    paintable_width = canvas_width - 2 * line

    left_w = paintable_width * 2 / 10
    middle_w = paintable_width * 4 / 10
    right_w = paintable_width * 4 / 10

    left = panel(left_w, canvas_height, line, False)
    middle = panel(middle_w, canvas_height, line, False)
    right = panel(right_w, canvas_height, line, True)

    return beside(
        beside(
            beside(left, vertical_line(canvas_height, line)),
            middle
        ),
        beside(
            vertical_line(canvas_height, line),
            right
        )
    )

show_graphic(composition(400, 800, 10))
```
