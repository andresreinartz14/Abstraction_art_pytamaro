Welcome to **Abstraction in Art and Programming**.

You'll meet a big idea that artists and programmers both use:

> **Abstraction means choosing what matters, and leaving out what does not.**

That sounds serious, but it is actually something you already do all the time.

A map leaves out every blade of grass.
An emoji leaves out every wrinkle on a face.
A game icon leaves out the whole real object and keeps only the part your brain needs.

Artists do this. Programmers do this too.


## The Core Question

**How do artists and programmers both simplify reality?**

An abstract artist does not copy every detail of the world.
A programmer does not write out every tiny step every time.

Instead, both ask:

- What is essential?
- What structure should stay?
- What can be simplified without looosing the main idea?

That shared way of thinking is the heart of this course.

## Abstraction in Art

The goal is not to make the image "random" or "messy".
The goal is often the opposite:
to make the important ideas easier to see through shapes and structure.


## Abstraction in Programming

Instead of paint, they use **code**.

They might:
- hide complexity inside a function,
- give a useful name to a repeated pattern,

For example, if you want the same shape several times, you should not write everything again and again if one named piece can do the job.

## Shared Mindset: Artist 🤝 Programmer

An artist and a programmer may look very different while working, but their thinking can be surprisingly similar.

Both of them:
- choose what matters,
- create with structure,
- build a system, not just a final result.

An artist might think in shapes and balance.
A programmer might think in functions and composition.

But both are asking:

> "What is the clearest way to build this idea?"

## Make the Program More Readable... Introducing, FUNCTIONS!

So far, we have created the image step by step, great right?!

But programmers often like to group a whole idea into <u>something</u> with a meaningful name.
That <u>something</u> is a **function**. 

A function is a named block of code that performs a task. See how powerful this could be? By naming a block of code, we can call that name ANYWHERE in the program and that block will be executed. Say bye to re-writing steps! Just call it by it's name. 

So let's put our code that produces a cross into a function.
