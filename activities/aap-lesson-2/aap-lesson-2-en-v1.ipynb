## From One Big Picture to a Small Visual Vocabulary

An artist does not always invent every square of a composition from scratch.
Often, they reuse a few visual ideas again and again.

A programmer does something similar with **helper functions**.

A helper function is a small named piece of code that does one clear job.

So instead of writing:

- rectangle here
- rectangle there
- another rectangle there
- and another
- and another
- and oh no why is this my whole evening?

we can name a reusable building block once, then call it whenever we need it.

## Starter code: colors, size, and imports

We will use a fixed cell size in this lesson.  
That keeps the focus on **decomposition and naming**, not on changing proportions yet.

You will learn much more about varying sizes with parameters in **Lesson 3**.

```python
from pytamaro import Graphic, rectangle, beside, above, show_graphic, rgb_color, Color, empty_graphic

# A small palette inspired by the artwork
YELLOW     = rgb_color(236, 212, 150)
RED        = rgb_color(200,  98,  86)
DARKRED    = rgb_color(170,  83,  73)
BLUE       = rgb_color( 72, 102, 176)
LIGHTBLUE  = rgb_color(138, 156, 200)
OFFWHITE   = rgb_color(241, 238, 231)
LIGHTGREY  = rgb_color(201, 197, 191)
MIDGREY    = rgb_color(145, 141, 138)
DARKGREY   = rgb_color(100,  98,  96)
CHARCOAL   = rgb_color( 88,  88,  88)

WIDTH = 150
HEIGHT = 80
```

## Building block 1: a full cell

A **full** cell is just one <API>rectangle</API> filling the whole space.

This is our easiest helper function:
it hides the drawing details behind a short, clear name.

```python
def full(c: Color) -> Graphic:
    return rectangle(WIDTH, HEIGHT, c)

show_graphic(full(YELLOW))
```

## Building block 2: a double cell

A **double** cell splits the space into a **top half** and a **bottom half**.

Notice what this helper function does for us:
we no longer have to remember the exact rectangle sizes every time.
The *name* tells the story.

<Task>
Complete the `double` helper below.

Hints:
- each rectangle uses the full width
- each one uses half the height
- use `above(...)` to stack the two halves
</Task>

```python
def double(top_c: Color, bottom_c: Color) -> Graphic:
    return above(
        rectangle(WIDTH, ..., top_c),
        rectangle(WIDTH, ..., bottom_c)
    )

show_graphic(double(RED, LIGHTGREY))
```

## Building block 3: a quad cell

A **quad** cell breaks one cell into **four smaller rectangles**.

Now we are not just stacking shapes.
We are building a structure from smaller structures.
That is abstraction at work.

<DoNow id="predict-width">
Before, for the **double** cell, each individual rectangle used the full width.
For a **quad** cell, what is the new width of each rectangle?

Say it out loud!
</DoNow>

It is half the full `WIDTH`. Each row now has two rectangles <API>beside</API> each other, meaning for the total width to be equal to `WIDTH`, each rectangle must be `WIDTH/2`. 

<DoNow id="predict-height">
Before, for the **double** cell, each individual rectangle used half the height.
For a **quad** cell, what is the new height of each rectangle?

Say it out loud!
</DoNow>
It stays the same, `HEIGHT/2`. The **quad** cell has the same number of rows (two) as the **double** cell.

<Task>
Complete the missing function calls so the cell is built from:

- a top row with `c1` and `c2`
- a bottom row with `c3` and `c4`
</Task>

```python
def quad(c1: Color, c2: Color, c3: Color, c4: Color) -> Graphic:
    top_row = beside(
        rectangle(WIDTH / 2, HEIGHT / 2, c1),
        rectangle(WIDTH / 2, HEIGHT / 2, c2)
    )

    bottom_row = beside(
        rectangle(WIDTH / 2, HEIGHT / 2, c3),
        rectangle(WIDTH / 2, HEIGHT / 2, c4)
    )

    return ...(top_row, bottom_row)

show_graphic(quad(YELLOW, DARKGREY, OFFWHITE, RED))
```

## Building block 4: a more detailed cell

Some cells in the painting are more subdivided.
We will use a helper called `oct` for that.

You do **not** need to invent this one from scratch today.
Instead, read it like a detective:

- How wide is each small rectangle?
- Why are there 8 colors?
- How is the top half combined?
- How is the bottom half combined?

Reading structure is just as important as writing it.

```python
def oct(
    c1: Color, c2: Color, c3: Color, c4: Color,
    c5: Color, c6: Color, c7: Color, c8: Color
) -> Graphic:
    top = beside(
        beside(
            beside(
                rectangle(WIDTH / 4, HEIGHT / 2, c1),
                rectangle(WIDTH / 4, HEIGHT / 2, c2)
            ),
            rectangle(WIDTH / 4, HEIGHT / 2, c3)
        ),
        rectangle(WIDTH / 4, HEIGHT / 2, c4)
    )

    bottom = beside(
        beside(
            beside(
                rectangle(WIDTH / 4, HEIGHT / 2, c5),
                rectangle(WIDTH / 4, HEIGHT / 2, c6)
            ),
            rectangle(WIDTH / 4, HEIGHT / 2, c7)
        ),
        rectangle(WIDTH / 4, HEIGHT / 2, c8)
    )

    return above(top, bottom)

show_graphic(oct(RED, RED, BLUE, RED, LIGHTBLUE, LIGHTGREY, CHARCOAL, MIDGREY))
```

## From cells to rows

Once we have cell types, we can stop thinking about lonely rectangles.
Now we can think in larger units:

- cells make a **row**
- rows make a **grid**
- the grid makes the **artwork**

That is “parts of a whole” thinking.

We will use two helper functions, `row` and `grid`.

These functions use a `for` loop to combine many graphics.
Don't worry if loops still feel a bit mysterious right now — **Lesson 4** will explore repetition in much more detail.

For this lesson, focus on the big idea:

> helper functions let us build a large picture from reusable pieces.

```python
def row(cells: list[Graphic]) -> Graphic:
    result = empty_graphic()
    for g in cells:
        result = beside(result, g)
    return result


def grid(rows: list[list[Graphic]]) -> Graphic:
    result = empty_graphic()
    for r in rows:
        result = above(result, row(r))
    return result
```

## Rebuilding the artwork as a grid of ideas

Here comes the coolest part.

We are **not** drawing every rectangle manually.

We are describing the artwork as a **2D list of chosen cell types**.
That means the painting becomes something like a sentence made from a small vocabulary.

![grid-plan](colorList.png)

<Task>
Complete the missing helper calls in the artwork plan below.

Tips:

- use `full(...)` when the whole cell is one color
- use `double(...)` for two stacked halves
- use `quad(...)` for four quarter blocks

```python
art = [
    [
        full(YELLOW),
        oct(RED, RED, BLUE, RED, LIGHTBLUE, LIGHTGREY, CHARCOAL, MIDGREY),
        quad(YELLOW, DARKGREY, OFFWHITE, RED)
    ],
    [
        ...(OFFWHITE),              # one solid cell
        ...(RED, LIGHTGREY),        # top and bottom
        ...(MIDGREY, DARKGREY)      # top and bottom
    ],
    [
        ...(CHARCOAL, LIGHTGREY, RED, RED),  # four parts
        double(DARKGREY, YELLOW),
        double(RED, DARKRED)
    ]
]

show_graphic(grid(art))
```

## Why this matters

Imagine writing the whole painting without helper functions.

You would need to remember:

- every rectangle
- every size
- every position
- every color
- and probably your emotional support snack

With helper functions, we can describe the artwork at a **higher level**:

- “this cell is a `quad`”
- “this one is a `double`”
- “this row has three cells”

That makes the code shorter, clearer, and easier to change.

```python
# Complete reference version

def double(top_c: Color, bottom_c: Color) -> Graphic:
    return above(
        rectangle(WIDTH, HEIGHT / 2, top_c),
        rectangle(WIDTH, HEIGHT / 2, bottom_c)
    )

def quad(c1: Color, c2: Color, c3: Color, c4: Color) -> Graphic:
    top_row = beside(
        rectangle(WIDTH / 2, HEIGHT / 2, c1),
        rectangle(WIDTH / 2, HEIGHT / 2, c2)
    )
    bottom_row = beside(
        rectangle(WIDTH / 2, HEIGHT / 2, c3),
        rectangle(WIDTH / 2, HEIGHT / 2, c4)
    )
    return above(top_row, bottom_row)

art = [
    [
        full(YELLOW),
        oct(RED, RED, BLUE, RED, LIGHTBLUE, LIGHTGREY, CHARCOAL, MIDGREY),
        quad(YELLOW, DARKGREY, OFFWHITE, RED)
    ],
    [
        full(OFFWHITE),
        double(RED, LIGHTGREY),
        double(MIDGREY, DARKGREY)
    ],
    [
        quad(CHARCOAL, LIGHTGREY, RED, RED),
        double(DARKGREY, YELLOW),
        double(RED, DARKRED)
    ]
]

show_graphic(grid(art))
```
